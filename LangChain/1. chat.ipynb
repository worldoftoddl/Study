#### 3. Anthropic 랭체인 (불친절한 에러코드 - 공식문서 참조 必)

```python
from langchain_anthropic import ChatAnthropic

llm = ChatAnthropic(
    model="claude-3-haiku-20240307",
    temperature=0,
    max_tokens=200, 
    )
```

```python
llm.invoke("What if the capital city of Republic of Korea?")
```
