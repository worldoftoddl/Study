```python
llm.invoke('What is the capital city of Korea?')
```

#### 1. PromptValue

```python
'''
PromptValue는 LangChain의 PromptTemplate로 만들 수 있음
'''
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
  template='What is the capital of {country}?',    ## {} 중괄호 안에는 placeholder - 나중에 값이 들어갈 자리
  input_variables = ["country"]                    ## plceholder들의 리스트 - 최신 LangChain에서는 생략해도 인식함
)

prompt = prompt_template.invoke({"country": "Republic of Korea"})

print(prompt)   ## 이 변수 prompt가 PromptValue
```

```python
llm.invoke(prompt)
```

#### 2. BaseMessage의 리스트

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

'''
A `HumanMessage` is a message that is passed in from a user to the model.

Example:
        from langchain_core.messages import HumanMessage, SystemMessage

        messages = [
            SystemMessage(content="You are a helpful assistant! Your name is Bob."),
            HumanMessage(content="What is your name?"),
        ]

        # Instantiate a chat model and invoke it with the messages
        model = ...
        print(model.invoke(messages))
'''

## 여기서 SystemMessage는 LLM의 페르소나,
## HumanMessage는 인간(user)가 LLM에게 전달하는 메시지

## 이러한 Message들의 list를 llm.invoke()에 담아 전달할 수 있음!
```

```python
basemessage_list = [
  SystemMessage(content='당신은 친절하고 전문적인 회계사입니다'), # LLM의 페르소나를 제어
  HumanMessage(content='복구충당부채에 대해 설명해주세요')
]

response = llm.invoke(basemessage_list)

print(response)
print('\n', type(response), sep='')   # llm.invoke()는 AIMessage를 반환함
```

#### 3. BaseMessage를 통한 context 유지

```python
basemessage_list.append(response)
basemessage_list.append(HumanMessage(content='간단한 예시를 들어줘'))

response2 = llm.invoke(basemessage_list)

print(response2)
## 이런 방식으로 대화 히스토리를 쌓아 맥락을 유지시킬 수 있음
```

```python
'''
즉, 정리를 해보자면, 이 방식을 사용할 때 LLM들은 
user가 질문을 할 때마다 이전의 모든 대화를 한번씩 읽고 와서 답변을 함

그래서 대화 길이에 한계(context window)가 있는 것이고,
그 한계를 넘어서게 되면 앞선 맥락이 손실되는 것임. 

나중에는 이 한계를 극복하기 위해 Memory 기능을 사용하게 될 것임!

ConversationBufferMemory()        # 전부 저장 (기본)
ConversationBufferWindowMemory(k=5)  # 최근 k개만
ConversationSummaryMemory()       # 요약해서 저장
'''
```

#### 4. LangChain에서 사용할 방법은?

```python
## 그러나 앞으로 사용할 LCEL 방법에서는 앞서 알아본 BaseMessage의 프롬프트 템플릿이 아닌
## 아래 설명할 ChatPromptTemplate의 형식으로 context를 유지할 것임
```

```python
from langchain_core.prompts import ChatPromptTemplate


## 이런 방식으로 대화 context를 튜플 리스트 형식으로 전달하는 방식을 사용할 것(LCEL)
chat_prompt_template = ChatPromptTemplate.from_messages([
  ('system', '당신은 친절하고 전문적인 회계사입니다.'),
  ('human', '{subject}의 개념과 예시에 대해서 설명하시오')
])

chat_prompt = chat_prompt_template.invoke({'subject': "대손충당금"})
llm.invoke(chat_prompt)

# 이 부분을 LCEL에서는
#  chain = chat_prompt_template | llm | StrOutputParser()
#  chai.invoke({'subject': '대손충당금'})
# 이런 방식으로 사용하게 됨 (앞으로 배울겨!)
```
